### Proposition — Charge Conservation from Maxwell Equations with Sources

Let Maxwell's equations with sources be

$$
\nabla\cdot\vec{E}
=
\frac{\rho}{\epsilon_0}
$$

and

$$
\nabla\times\vec{B}
=
\mu_0\vec{J}
+
\mu_0\epsilon_0
\frac{\partial \vec{E}}{\partial t}.
$$

Then the four-current

$$
J^\mu=(c\rho,\vec{J})
$$

satisfies the continuity equation

$$
\partial_\mu J^\mu=0.
$$

<!-- proof:start -->
### Proof

Start from Ampère-Maxwell law:

$$
\nabla\times\vec{B}
=
\mu_0\vec{J}
+
\mu_0\epsilon_0
\frac{\partial \vec{E}}{\partial t}.
$$

Take the divergence of both sides:

$$
\nabla\cdot(\nabla\times\vec{B})
=
\nabla\cdot
\left(
\mu_0\vec{J}
+
\mu_0\epsilon_0
\frac{\partial \vec{E}}{\partial t}
\right).
$$

Using

$$
\nabla\cdot(\nabla\times\vec{B})=0,
$$

we get

$$
0
=
\mu_0\nabla\cdot\vec{J}
+
\mu_0\epsilon_0
\frac{\partial}{\partial t}
(\nabla\cdot\vec{E}).
$$

Now use Gauss's law:

$$
\nabla\cdot\vec{E}
=
\frac{\rho}{\epsilon_0}.
$$

Therefore,

$$
\frac{\partial}{\partial t}
(\nabla\cdot\vec{E})
=
\frac{\partial}{\partial t}
\left(
\frac{\rho}{\epsilon_0}
\right)
=
\frac{1}{\epsilon_0}
\frac{\partial \rho}{\partial t}.
$$

Substituting,

$$
0
=
\mu_0\nabla\cdot\vec{J}
+
\mu_0\epsilon_0
\left(
\frac{1}{\epsilon_0}
\frac{\partial \rho}{\partial t}
\right).
$$

Canceling $\epsilon_0$,

$$
0
=
\mu_0\nabla\cdot\vec{J}
+
\mu_0
\frac{\partial \rho}{\partial t}.
$$

Dividing by $\mu_0$,

$$
\nabla\cdot\vec{J}
+
\frac{\partial \rho}{\partial t}
=
0.
$$

Thus,

$$
\frac{\partial \rho}{\partial t}
+
\nabla\cdot\vec{J}
=
0.
$$

Now write the four-current as

$$
J^\mu=(c\rho,\vec{J})
$$

and the covariant derivative as

$$
\partial_\mu
=
\left(
\frac{1}{c}\frac{\partial}{\partial t},
\nabla
\right).
$$

Then

$$
\partial_\mu J^\mu
=
\frac{1}{c}\frac{\partial}{\partial t}(c\rho)
+
\nabla\cdot\vec{J}.
$$

Therefore,

$$
\partial_\mu J^\mu
=
\frac{\partial \rho}{\partial t}
+
\nabla\cdot\vec{J}.
$$

But we already proved that

$$
\frac{\partial \rho}{\partial t}
+
\nabla\cdot\vec{J}
=
0.
$$

Hence,

$$
\partial_\mu J^\mu=0.
$$
<!-- proof:end -->
