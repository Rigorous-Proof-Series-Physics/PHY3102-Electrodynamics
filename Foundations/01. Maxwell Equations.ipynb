### Proposition — Electric Field from Scalar and Vector Potentials

Let

$$
V(\mathbf{r},t)
$$

be the scalar potential and

$$
\mathbf{A}(\mathbf{r},t)
$$

the vector potential associated with an electromagnetic field.

Then the electric field is given by

$$
\mathbf{E}
=
-\nabla V
-
\frac{\partial \mathbf{A}}{\partial t}.
$$

<!-- proof:start -->
From Maxwell–Faraday equation,

$$
\nabla \times \mathbf{E}
=
-
\frac{\partial \mathbf{B}}{\partial t},
$$

introduce a vector potential

$$
\mathbf{B}
=
\nabla \times \mathbf{A}.
$$

Substituting into Faraday’s law yields

$$
\nabla \times \mathbf{E}
=
-
\frac{\partial}{\partial t}
(\nabla \times \mathbf{A}).
$$

Using the linearity of the curl operator,

$$
\nabla \times
\left(
\mathbf{E}
+
\frac{\partial \mathbf{A}}{\partial t}
\right)
=
0.
$$

A vector field with vanishing curl is conservative, therefore there exists a scalar function

$$
V(\mathbf{r},t)
$$

such that

$$
\mathbf{E}
+
\frac{\partial \mathbf{A}}{\partial t}
=
-
\nabla V.
$$

Hence,

$$
\mathbf{E}
=
-\nabla V
-
\frac{\partial \mathbf{A}}{\partial t}.
$$

### Proposition — Magnetic Field from the Vector Potential

Let

$$
\mathbf{B}(\mathbf{r},t)
$$

be a magnetic field satisfying

$$
\nabla \cdot \mathbf{B}=0.
$$

Then there exists a vector potential

$$
\mathbf{A}(\mathbf{r},t)
$$

such that

$$
\mathbf{B}
=
\nabla \times \mathbf{A}.
$$

<!-- proof:start -->
One of Maxwell’s equations states that

$$
\nabla \cdot \mathbf{B}=0.
$$

A fundamental identity from vector calculus is

$$
\nabla \cdot (\nabla \times \mathbf{A})=0
$$

for every sufficiently smooth vector field

$$
\mathbf{A}.
$$

Therefore any field written as a curl automatically satisfies the divergence-free condition.

Conversely, in a simply connected region, every smooth vector field with zero divergence can be represented as the curl of another vector field. Hence there exists a vector potential

$$
\mathbf{A}
$$

such that

$$
\mathbf{B}
=
\nabla \times \mathbf{A}.
$$
<!-- block:end -->

### Proposition — Electric Field Written in Terms of Potentials

Let the magnetic field be represented by a vector potential $\vec{A}$:

$$
\vec{B}=\nabla\times\vec{A}.
$$

Then Faraday's law,

$$
\nabla\times\vec{E}
=
-\frac{\partial \vec{B}}{\partial t},
$$

implies that the electric field can be written as

$$
\vec{E}
=
-\nabla\phi
-
\frac{\partial \vec{A}}{\partial t},
$$

for some scalar potential $\phi$.

<!-- proof:start -->
### Proof

Start from Faraday's law:

$$
\nabla\times\vec{E}
=
-\frac{\partial \vec{B}}{\partial t}.
$$

Using

$$
\vec{B}
=
\nabla\times\vec{A},
$$

we get

$$
\nabla\times\vec{E}
=
-\frac{\partial}{\partial t}
\left(
\nabla\times\vec{A}
\right).
$$

Since the time derivative commutes with the curl,

$$
\nabla\times\vec{E}
=
-\nabla\times
\frac{\partial\vec{A}}{\partial t}.
$$

Move everything to the same side:

$$
\nabla\times\vec{E}
+
\nabla\times
\frac{\partial\vec{A}}{\partial t}
=
0.
$$

Using linearity of the curl,

$$
\nabla\times
\left(
\vec{E}
+
\frac{\partial\vec{A}}{\partial t}
\right)
=
0.
$$

Therefore, the vector field

$$
\vec{E}
+
\frac{\partial\vec{A}}{\partial t}
$$

is irrotational.

Hence, locally, it can be written as the gradient of a scalar function. We define that scalar function as $-\phi$:

$$
\vec{E}
+
\frac{\partial\vec{A}}{\partial t}
=
-\nabla\phi.
$$

Therefore,

$$
\vec{E}
=
-\nabla\phi
-
\frac{\partial\vec{A}}{\partial t}.
$$
<!-- proof:end -->

### Proposition — Existence of the Vector Potential

Let the magnetic field satisfy Gauss's law for magnetism:

$$
\nabla\cdot\vec{B}=0.
$$

Then there exists a vector field $\vec{A}$ such that

$$
\vec{B}
=
\nabla\times\vec{A}.
$$

<!-- proof:start -->
### Proof

Start from Maxwell's equation for the magnetic field:

$$
\nabla\cdot\vec{B}=0.
$$

A vector field with zero divergence is called solenoidal.

Now recall the vector calculus identity

$$
\nabla\cdot(\nabla\times\vec{F})=0
$$

for every sufficiently smooth vector field $\vec{F}$.

Therefore, every curl automatically has zero divergence.

The converse statement from vector calculus (Poincaré lemma in simply connected regions) states that if a vector field has zero divergence, then locally it can be represented as the curl of another vector field.

Hence, since

$$
\nabla\cdot\vec{B}=0,
$$

there exists a vector field $\vec{A}$ such that

$$
\vec{B}
=
\nabla\times\vec{A}.
$$

The field $\vec{A}$ is called the vector potential.

Finally, substituting this expression into Gauss's law for magnetism gives

$$
\nabla\cdot\vec{B}
=
\nabla\cdot(\nabla\times\vec{A})
=
0,
$$

confirming consistency with Maxwell's equations.
<!-- proof:end -->
