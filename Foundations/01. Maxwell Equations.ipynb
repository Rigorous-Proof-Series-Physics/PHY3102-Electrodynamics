### Proposition — Electric Field from Scalar and Vector Potentials

Let

$$
V(\mathbf{r},t)
$$

be the scalar potential and

$$
\mathbf{A}(\mathbf{r},t)
$$

the vector potential associated with an electromagnetic field.

Then the electric field is given by

$$
\mathbf{E}
=
-\nabla V
-
\frac{\partial \mathbf{A}}{\partial t}.
$$

<!-- proof:start -->
From Maxwell–Faraday equation,

$$
\nabla \times \mathbf{E}
=
-
\frac{\partial \mathbf{B}}{\partial t},
$$

introduce a vector potential

$$
\mathbf{B}
=
\nabla \times \mathbf{A}.
$$

Substituting into Faraday’s law yields

$$
\nabla \times \mathbf{E}
=
-
\frac{\partial}{\partial t}
(\nabla \times \mathbf{A}).
$$

Using the linearity of the curl operator,

$$
\nabla \times
\left(
\mathbf{E}
+
\frac{\partial \mathbf{A}}{\partial t}
\right)
=
0.
$$

A vector field with vanishing curl is conservative, therefore there exists a scalar function

$$
V(\mathbf{r},t)
$$

such that

$$
\mathbf{E}
+
\frac{\partial \mathbf{A}}{\partial t}
=
-
\nabla V.
$$

Hence,

$$
\mathbf{E}
=
-\nabla V
-
\frac{\partial \mathbf{A}}{\partial t}.
$$

### Proposition — Magnetic Field from the Vector Potential

Let

$$
\mathbf{B}(\mathbf{r},t)
$$

be a magnetic field satisfying

$$
\nabla \cdot \mathbf{B}=0.
$$

Then there exists a vector potential

$$
\mathbf{A}(\mathbf{r},t)
$$

such that

$$
\mathbf{B}
=
\nabla \times \mathbf{A}.
$$

<!-- proof:start -->
One of Maxwell’s equations states that

$$
\nabla \cdot \mathbf{B}=0.
$$

A fundamental identity from vector calculus is

$$
\nabla \cdot (\nabla \times \mathbf{A})=0
$$

for every sufficiently smooth vector field

$$
\mathbf{A}.
$$

Therefore any field written as a curl automatically satisfies the divergence-free condition.

Conversely, in a simply connected region, every smooth vector field with zero divergence can be represented as the curl of another vector field. Hence there exists a vector potential

$$
\mathbf{A}
$$

such that

$$
\mathbf{B}
=
\nabla \times \mathbf{A}.
$$
