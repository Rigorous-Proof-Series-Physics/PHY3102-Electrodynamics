### Proposition — Wave Equation for the Scalar Potential in the Lorenz Gauge

Let the electromagnetic potentials be defined by

$$
\vec{B}=\nabla\times \vec{A}
$$

and

$$
\vec{E}=-\nabla\phi-\frac{\partial \vec{A}}{\partial t}.
$$

Assume the Lorenz gauge condition

$$
\nabla\cdot\vec{A}
+
\frac{1}{c^2}\frac{\partial \phi}{\partial t}
=
0.
$$

Then the scalar potential $\phi$ satisfies

$$
\nabla^2\phi
-
\frac{1}{c^2}\frac{\partial^2\phi}{\partial t^2}
=
-\frac{\rho}{\epsilon_0}.
$$

<!-- proof:start -->
### Proof

Start from Gauss's law with source:

$$
\nabla\cdot\vec{E}
=
\frac{\rho}{\epsilon_0}.
$$

Substitute the potential expression for the electric field:

$$
\vec{E}
=
-\nabla\phi
-
\frac{\partial \vec{A}}{\partial t}.
$$

Therefore,

$$
\nabla\cdot\vec{E}
=
\nabla\cdot
\left(
-\nabla\phi
-
\frac{\partial \vec{A}}{\partial t}
\right).
$$

Using linearity of the divergence,

$$
\nabla\cdot\vec{E}
=
-\nabla\cdot\nabla\phi
-
\nabla\cdot\frac{\partial \vec{A}}{\partial t}.
$$

Since

$$
\nabla\cdot\nabla\phi
=
\nabla^2\phi,
$$

and assuming the spatial divergence commutes with the time derivative,

$$
\nabla\cdot\frac{\partial \vec{A}}{\partial t}
=
\frac{\partial}{\partial t}
\left(
\nabla\cdot\vec{A}
\right),
$$

we get

$$
\nabla\cdot\vec{E}
=
-\nabla^2\phi
-
\frac{\partial}{\partial t}
\left(
\nabla\cdot\vec{A}
\right).
$$

By Gauss's law,

$$
-\nabla^2\phi
-
\frac{\partial}{\partial t}
\left(
\nabla\cdot\vec{A}
\right)
=
\frac{\rho}{\epsilon_0}.
$$

Now use the Lorenz gauge condition:

$$
\nabla\cdot\vec{A}
+
\frac{1}{c^2}\frac{\partial \phi}{\partial t}
=
0.
$$

Hence,

$$
\nabla\cdot\vec{A}
=
-
\frac{1}{c^2}
\frac{\partial \phi}{\partial t}.
$$

Taking a time derivative,

$$
\frac{\partial}{\partial t}
\left(
\nabla\cdot\vec{A}
\right)
=
-
\frac{1}{c^2}
\frac{\partial^2\phi}{\partial t^2}.
$$

Substitute this into Gauss's law:

$$
-\nabla^2\phi
-
\left(
-
\frac{1}{c^2}
\frac{\partial^2\phi}{\partial t^2}
\right)
=
\frac{\rho}{\epsilon_0}.
$$

Thus,

$$
-\nabla^2\phi
+
\frac{1}{c^2}
\frac{\partial^2\phi}{\partial t^2}
=
\frac{\rho}{\epsilon_0}.
$$

Multiplying both sides by $-1$, we obtain

$$
\nabla^2\phi
-
\frac{1}{c^2}
\frac{\partial^2\phi}{\partial t^2}
=
-
\frac{\rho}{\epsilon_0}.
$$

Therefore, in the Lorenz gauge, the scalar potential obeys the inhomogeneous wave equation

$$
\nabla^2\phi
-
\frac{1}{c^2}
\frac{\partial^2\phi}{\partial t^2}
=
-
\frac{\rho}{\epsilon_0}.
$$
